### learning the slicing of strig 


```python
file = "myfile.pdf"

dot = file.index('.');

print(dot)
print("extension : " , file[ dot :  : 1])
print("file name : " , file[ : dot  : 1])
```

## learning multipal build in strign methods 

```python
name = "  aditya mohod  "

print(name.upper())                 # Converts all letters to uppercase

print(name.lower())                 # Converts all letters to lowercase

print(name.capitalize())            # Capitalizes the first character

print(name.title())                 # Capitalizes the first letter of every word

print(name.strip())                 # Removes spaces from beginning and end

print(name.replace("aditya", "aadi")) # Replaces "aditya" with "aadi"

print(name.split())                 # Splits the string into a list of words

print("-".join(["hello", "world"])) # Joins list elements using "-"

print(name.find("m"))               # Returns the index of "m"

print(name.count("o"))              # Counts how many times "o" appears

print(name.startswith("aditya"))       # Checks if string starts with "aditya"

print(name.endswith("mohod"))          # Checks if string ends with "mohod"

print(name.isalpha())                  # Checks if all characters are alphabets
                                       # False because there is a space

print(name.isalnum())                  # Checks if all characters are letters/numbers
                                       # False because there is a space

print(name.islower())                  # Checks if all letters are lowercase

print(name.isupper())                  # Checks if all letters are uppercase

print(name.istitle())                  # Checks if string is in Title Case

print(name.isspace())                  # Checks if string contains only spaces

print(name.swapcase())                 # Converts lowercase to uppercase and vice versa

print(name.center(30))                 # Places the string in the center of 30 characters

print(name.ljust(30))                  # Left-aligns the string within 30 characters

print(name.rjust(30))                  # Right-aligns the string within 30 characters

print(name.zfill(20))                  # Adds zeros to the beginning
```
