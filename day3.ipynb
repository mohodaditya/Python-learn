```python
#this program filtes the list data based on the data type it only returns the str datatype values 

names = ["aditya", "yash", "om", "ram", "sham", 132, ["yash", 12, 13.09]]
filtered_list = []

print(names)

for name in names :
    if isinstance(name, str):
        filtered_list.append(name)

print(filtered_list)
```

```python

names = ['aditya', 'yash', 'om', 'ram', 'sham', 'jayesh', 'rajiv', 'tukaram']
new = []

for name in names:
    if len(name)>3:
        print(name)
        new.append(name)
```

```python
names = ['aditya', 'yash', 'om', 'ram', 'sham', 'jayesh', 'rajiv', 'tukaram']
new1 = []

for name in names:
    new1.append(name[::-1])
print(new1)
```

```python
for name in names:
    for n in name:
        if n =='m':
            print(name)
```

```python
for name in names:
    for n in name:
        if name[0]=='r':
            print(name)
print(names)
```

### --> more list methods in python 


```python
print(names)

names.append("ayush") # .append(element) adds the single element at the last 
print(names)

names.extend(["uday", "ritesh"]) # .extend(ele1, ele2) adds the multipal elements to the list at the last 
print(names)


names.insert(1, "yash") # .insert(index, element) aads the element in the list on a spacific index anywhere in the list it takes a index & element as an input 
print(names)

names.remove("ayush") # .remove(element) it removes the first matching element from the list 
print(names)


item = names.pop(1)  # .pop(index) removes the element based on the given index and also returns it 
print(item)

names.pop() # .pop() if the index parmeter is not pass to the pop method it will simple removes the last element 
print(names)

names.clear() # .clear() method is removes the each and every element from the list 
print(names)

names.extend(['aditya', 'yash', 'om', 'ram', 'sham', 'jayesh', 'rajiv', 'tukaram', 'ayush', 'uday', 'ritesh'])
print(names)

print(names.index("tukaram")) # .index(element) returns the index of the given element 

print(names.count("aditya")) # .count(element) returns the count/requency of the element in the list

names.sort() # .sort() sorts the elements in the list in acending order 
print(names)

names.sort(reverse=True) # .sort(reverse=True) sorts the elements from the list according to the decendeing orders 
print(names)

names.reverse() # .reverse() method reverse the list means change the positions of the elements 1st goes last and last comes 1st
print(names)

new_names = names.copy() # .copy() returns a copy of orignal list 
print(new_names)

print(names)

```
