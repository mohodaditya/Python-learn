```python
numbers = [10,-20,30,-40,-50,60]

positive_num = []
negative_num = []

for n in numbers:
    if n<0:
        positive_num.append(n)
    else:
        negative_num.append(n)

print("these are numbers : ", numbers)
print("these are positive numbers : ", positive_num)
print("these are nrgaive numbers : ", negative_num)
```

```python
pro = [10000, 3000, 5000, 15000, 18000, 120000, 3000, 9000, 20000]
sell = []

for p in pro:
    if p>15000:
        sell.append(p-(p*15/100))
        print(p, "15 percent discount")
    elif p>10000:
        sell.append(p-(p*10/100))
        print(p, "10 percet discount ")
    elif p>5000:
        sell.append(p-(p*5/100))
        print(p, "5 percet dicsount")
    else:
        print("no discount")
    

print(sell)
```
